```python
import numpy as np
import zarr

data = np.arange(64*64).reshape(64,64)
```

```python
grp = zarr.create_group(store="good_or_bad.zarr")
chunks = grp.create_group(name="chunks")

bad  = chunks.create_array(name="bad", shape=(64, 64), chunks=(1, 1), dtype='int32')
good = chunks.create_array(name="good", shape=(64, 64), chunks=(64, 64), dtype='int32')
no_compression = chunks.create_array(name="no_compression", shape=(64, 64), chunks=(64, 64), dtype='int32', compressors=())

good[:] = data
bad[:] = data
no_compression[:] = data
```

What's wrong with `bad.zarr`?

```python
chunksAgain = grp.create_group(name="chunksAgain")

good = chunksAgain.create_array(name="good", shape=(64, 64), chunks=(64, 64), dtype='int32')
bad = chunksAgain.create_array(name="bad", shape=(64, 64), chunks=(256, 256), dtype='int32', compressors=())
silly = chunksAgain.create_array(name="silly", shape=(64, 64), chunks=(256, 256), dtype='int32')

good[:] = data
bad[:] = data
silly[:] = data
```
