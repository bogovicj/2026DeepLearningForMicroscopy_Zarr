```python
import numpy as np
import zarr
```

```python
grp = zarr.create_group(store="memory-layout.zarr")
monolith = grp.create_array(name="monolith", shape=(2,3,5), chunks=(2,3,5), dtype='uint8', compressors=())
monolith[:] = np.arange(2*3*5).reshape(2,3,5)
monolith[:]
```

```python
!od -t d1 -w5 memory-layout.zarr/monolith/c/0/0/0
```

```python
zslices = grp.create_array(name="zslices", shape=(2,3,5), chunks=(1,3,5), dtype='uint8', compressors=())
zslices[:] = np.arange(2*3*5).reshape(2,3,5)
zslices[:]
```

```python
!od -t d1 -w5 memory-layout.zarr/zslices/c/0/0/0
```

```python
!od -t d1 -w5 memory-layout.zarr/zslices/c/1/0/0
```

```python
xrows = grp.create_array(name="xrows", shape=(2,3,5), chunks=(1,1,5), dtype='uint8', compressors=())
xrows[:] = np.arange(2*3*5).reshape(2,3,5)
xrows[:]
```

# Zarr chunks are always the same shape

Even when then extend past the boundaries of the image

```python
padded = grp.create_array(name="padded", shape=(2,3,5), chunks=(2,3,16), dtype='uint8', compressors=())
padded[:] = np.arange(2*3*5).reshape(2,3,5)
padded[:]
```

```python
!od -t d1 -w16 memory-layout.zarr/padded/c/0/0/0
```

## A more realistic example

```python
smallChunks = grp.create_array(name="smallChunks", shape=(2,3,5), chunks=(2,2,2), dtype='uint8', compressors=())
smallChunks[:] = np.arange(2*3*5).reshape(2,3,5)
smallChunks[:]
```

```python
!od -t d1 -w2 memory-layout.zarr/smallChunks/c/0/0/0
```
