```python
import tifffile
import numpy as np
from ome_zarr.writer import write_image
```

Hey! Don't use this for serious conversion of tiff files. It totally ignores important metadata. 

```python
# Get a sample image 
!curl https://imagej.net/ij/images/mitosis.tif > mitosis.tif
```

```python
image = tifffile.imread('mitosis.tif')
```

```python
# This image's shape is TZCYX
image.shape
```

```python
# Convert to TCZYX
imageTCZYX = np.permute_dims(image,[0,2,1,3,4])
imageTCZYX.shape
```

```python
# Write the image (with hard-coded resolutions and units)
write_image(
    imageTCZYX, 'mitosis.ome.zarr', 
    axes="tczyx",
    scale={"t": 0.14, "c" :1.0, "z": 1.0, "y": 0.0885000, "x": 0.0885000},
    axes_units={"t": "second", "c": "", "z": "micrometer", "y": "micrometer", "x": "micrometer"},
    )
```
